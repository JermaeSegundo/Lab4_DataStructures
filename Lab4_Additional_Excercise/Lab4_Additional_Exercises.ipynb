```python
# ENGINEERING TEST DATA ORGANIZER (DYNAMIC UNIQUE VERSION)

LAST_NAME = "SEGUNDO"
SEED_NUM = 2
FAVORITE_ARTIST = "LANY"

# 1. Generated Data
print("===== Generated Data =====")
test_sources = [f"{LAST_NAME} T{x}" for x in range(1, 6)]
print(f"Generated Test Data: {test_sources}")

# 2. Repeated/Distinct Data
print("\n===== Repeated/Distinct Data =====")
test_sources_with_repeat = test_sources + [test_sources[0]]
print(f"Test Data with Repetition: {test_sources_with_repeat}")

# 3. Organized Test Data
print("\n===== Organized Test Data =====")
categories = ["TEMPERATURE", "VOLTAGE", "CURRENT", "PRESSURE", "SPEED"]
for cat in categories:
    print(cat)

# Measurement
def get_record(i):
    seed_offset = SEED_NUM * 3
    return {
        "TEMPERATURE": f"{15 * i + seed_offset} C, {38.5 * i + SEED_NUM * 1.25} C",
        "VOLTAGE": f"{16 * i + seed_offset} V, {39.25 * i + SEED_NUM * 1.5} V",
        "CURRENT": f"{round((3.0 + i * 0.3) * SEED_NUM, 2)} A, {round((5.0 + i * 0.4) * SEED_NUM, 2)} A",
        "PRESSURE": f"{round((15.0 + i * 1.0) * SEED_NUM, 1)} atm, {round((37.5 + i * 1.2) * SEED_NUM, 1)} atm",
        "SPEED": f"{round((17.0 + i * 1.6) * SEED_NUM, 0)} km/hr, {round((40.25 + i * 0.9) * SEED_NUM, 4)} km/hr"
    }

# 4. Processed Results (Loops T1 to T5, plus repeated T1 at the end)
print("\n===== Processed Results =====")
sequence = [1, 2, 3, 4, 5, 1]
for idx in sequence:
    print(f"{LAST_NAME} T{idx}")
    record = get_record(idx)
    for cat, val in record.items():
        print(f"{cat}: {val}")

# 5. Ordered Results and Final Summary
print("\n===== Ordered Results and Final Summary =====")
for idx in range(1, 6):
    print(f"{LAST_NAME} T{idx}")
    record = get_record(idx)
    for cat, val in record.items():
        print(f"{cat}: {val}")
print("\nFinal Organized and Summarized Test Dataset generated successfully.")
```

```python
# STUDENT PERFORMANCE DATA ANALYZER 

LAST_NAME = "SEGUNDO"
SEED_NUM = 2
FAVORITE_ARTIST = "LANY"

# 1. Generated Performance Records
print("===== 1. Generated Performance Records =====")
# Generate student labels using FAVORITE_ARTIST characters
student_labels = [FAVORITE_ARTIST[i % len(FAVORITE_ARTIST)] + f"_L{i+1}" for i in range(5)]
students = [f"{LAST_NAME} {label}" for label in student_labels]
subjects = ["PROGRAMMING", "MATHEMATICS", "ELECTRONICS", "PHYSICS", "ENGLISH"]

print(f"Generated Students: {students}")
print(f"Subjects: {subjects}")

# Generate scores using SEED_NUM and FAVORITE_ARTIST character positions[span_3](start_span)[span_3](end_span)
performance_records = []
for s_idx, student in enumerate(students):
    for sub_idx, subject in enumerate(subjects):
        char_val = ord(FAVORITE_ARTIST[(s_idx + sub_idx) % len(FAVORITE_ARTIST)])
        # Dynamic score formula bounded between 70 and 100
        score = 70 + ((char_val * SEED_NUM + s_idx * 7 + sub_idx * 5) % 30)
        performance_records.append({"student": student, "subject": subject, "score": score})

# Include one repeated student-subject record[span_4](start_span)[span_4](end_span)
repeated_record = performance_records[0].copy()
performance_records.append(repeated_record)

# Include one incomplete record with a missing score[span_5](start_span)[span_5](end_span)
incomplete_record = {"student": students[-1], "subject": subjects[-1], "score": None}
performance_records.append(incomplete_record)

for rec in performance_records[:6]: # Display sample of generated records
    print(rec)

# 2. Organized Student/Subject Data
print("\n===== 2. Organized Student/Subject Data =====")
organized_data = {}
for r in performance_records:
    stud = r["student"]
    if stud not in organized_data:
        organized_data[stud] = {}
    # Keep the first valid score if duplicate occurs
    if r["subject"] not in organized_data[stud] or r["score"] is not None:
        organized_data[stud][r["subject"]] = r["score"]

for stud, subs in organized_data.items():
    print(f"\nStudent: {stud}")
    for sub, sc in subs.items():
        if sc is None:
            print(f"  {sub:<15} Grade: MISSING")
        else:
            print(f"  {sub:<15} Grade: {sc}")

# 3. Repeated and Incomplete Records[span_6](start_span)[span_6](end_span)
print("\n===== 3. Repeated/Incomplete Records =====")
print(f"Incomplete Record Found -> Student: {incomplete_record['student']}, Subject: {incomplete_record['subject']}, Score: MISSING")
print(f"Repeated Record Found -> {repeated_record}")

# 4. Performance Categories[span_7](start_span)[span_7](end_span)[span_8](start_span)[span_8](end_span)
print("\n===== 4. Performance Categories =====")
print("A: Excellent Performance (90-100)")
print("B: Good Performance (80-89)")
print("C: Satisfactory Performance (75-79)")
print("D: Needs Improvement (70-74)")
print("F: Failing Status (<70)")

# 5. Student and Subject Summaries[span_9](start_span)[span_9](end_span)
print("\n===== 5. Student and Subject Summaries =====")
student_averages = {}
for stud, subs in organized_data.items():
    valid_scores = [sc for sc in subs.values() if sc is not None]
    avg = sum(valid_scores) / len(valid_scores) if valid_scores else 0.0
    student_averages[stud] = avg

for stud, avg in student_averages.items():
    print(f"{stud} -> Average Score: {avg:.2f}%")

# 6. Ordered Results[span_10](start_span)[span_10](end_span)
print("\n===== 6. Ordered Results =====")
ordered_students = sorted(student_averages.items(), key=lambda x: x[1], reverse=True)
for stud, avg in ordered_students:
    print(f"Ranked -> {stud}: {avg:.2f}%")

# 7. Final Student Performance Report[span_11](start_span)[span_11](end_span)
print("\n===== 7. Final Student Performance Report =====")
for stud, avg in ordered_students:
    if avg >= 90:
        bracket = "A -> Excellent Performance"
    elif avg >= 80:
        bracket = "B -> Good Performance"
    elif avg >= 75:
        bracket = "C -> Satisfactory Performance"
    elif avg >= 70:
        bracket = "D -> Needs Improvement"
    else:
        bracket = "F -> Failing Status"
    
    print(f"\n{stud}")
    print(f"Average Score: {avg:.2f}% | Grade Bracket: {bracket}")
    print("-" * 65)

print("\nFinal Student Performance Report successfully produced.")
```

```python
# TRANSACTION HISTORY ANALYZER 

LAST_NAME = "SEGUNDO"
SEED_NUM = 2
FAVORITE_ARTIST = "LANY"

# 1. Generated Transaction Records
print("===== Generated Transanction Records =====")
customers = [f"{LAST_NAME} C{i}" for i in range(1, 6)]
print(f"Generated Transaction Records: {customers}")

# 2. Repeated/Distinct Transactions
print("\n===== Repeated/Distinct Transanctions =====")
customers_with_repeat = customers + [customers[0]]
print(f"Generated Transaction Records with Repetition: {customers_with_repeat}")

# 3. Organized Transaction Data
print("\n===== Organized Transanction Data =====")
products_list = ["NOTEBOOK", "PEN", "USB", "HEADPHONES", "CALCULATOR"]
for prod in products_list:
    print(prod)

# 4. Repeated Records Auditing
print("\n===== Repeated Records Auditing =====")
print(f"Repeated Student-Subject Record: {customers[0]}")
print(f"Repeated Transanction: {SEED_NUM}")

# 5 & 6. Customer and Product Summaries
print("\n===== Customer and Product Summaries =====")

# Product catalog with unit prices
prices = {
    "Notebook": 25.75,
    "Pen": 15.25,
    "USB": 78.35,
    "Headphones": 500.45,
    "Calculator": 2500.15
}

customer_due_totals = {}

for c_idx, cust in enumerate(customers, start=1):
    print(f"\nCUSTOMER: {cust}")
    print(f"{'PRODUCT':<15} {'PRICE PER UNIT':<15} {'QUANTITY':<15} {'TOTAL PRICE'}")
    print("-" * 65)
    
    total_due = 0
    for p_idx, (prod, price) in enumerate(prices.items(), start=1):
        # Dynamic quantity calculation driven by SEED_NUM and your unique inputs
        qty = p_idx + c_idx + (SEED_NUM - 1)
        total_price = price * qty
        total_due += total_price
        print(f"{prod:<15} {price:<15.2f} {qty:<15} {total_price:,.2f}")
    
    customer_due_totals[cust] = total_due
    print(f"DUE AMOUNT{'':<40} {total_due:,.2f}")

# 7 & 8. Final Consolidated Transaction Report
print("\n===== Final Consolidated Transaction Report =====")
for cust, due in customer_due_totals.items():
    print(f"{cust:<15} Total Due Amount: PHP {due:,.2f}")

print("\nFinal Consolidated Transaction Report successfully produced.")
```
